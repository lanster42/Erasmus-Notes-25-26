CORRECTNESS ANALYSIS:

Above are the recurrence equations for this problem. The idea is that for every $j \in {1, \ldots, n}$ and for every $k \in {1, \ldots, j}$ we can check if the word $s[k], \ldots, s[j]$ is in the dictionary and if it is, then recursively check whether there exists a solution (sequence of valid words) in $s[1], \ldots, s[k - 1]$. If such a solution exists, the function returns true, if it doesn't, it keeps looking until k = j and until j = n.

In implementation, this function would run until it received the first True (and also return it) or until it traverses all possibilities, after which the function woud return False.

COMPLEXITY ANALYSIS:

For every j we're checking whether the function yields true j times (because $k \in {1, \ldots, j}$ and because $d(w)$ takes constant time). In worst case scenarion this means that complexity is $O(n^2)$.
