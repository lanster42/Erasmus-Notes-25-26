CORRECTNESS ANALYSIS:

Above are the recurrence equations for this problem. The idea is that for every $j \in {1, \ldots, n}$ and for every $k \in {1, \ldots, j}$ we can check if the word $s[k], \ldots, s[j]$ is in the dictionary and if it is, then recursively check whether there exists a solution (sequence of valid words) in $s[1], \ldots, s[k - 1]$. If such a solution exists, the function returns true, if it doesn't, it keeps looking until k = j and until j = n.

In implementation, this function would run until it received the first True (and also return it) or until it traverses all possibilities, after which the function woud return False.

COMPLEXITY ANALYSIS:

For every j we're checking whether the function yields true j times (because $k \in {1, \ldots, j}$ and because $d(w)$ takes constant time). In worst case scenarion this means that complexity is $O(n^2)$.

EXPLANATION and CORRECTNESS ANALYSIS:

We're traversing the pattern p and the string w. When we encounter a match (w[i] = p[j]), we can move ahead and recursively compare the next 2 characters. If we encounter "?" in the pattern, we take either the next character in the pattern and the same character in the string (which would mean that the "?" replaced an empty character) or the next characters in both the string and the pattern (which would mean that the "?" replaced a non-empty character). If we encounter "\*", the first two possibilities are the same as in "?", but we get another possibiity, which is to take the next character in the sequence and the same character in the pattern ("\*"), which would mean that the "\*" replaced more than 1 character. The base cases return True if both i and j are 0 and false otherwise.

To get the solution, we could build a tree while making these choices, that is each match would build a node. In the end, we would just count the number of "true" leaves. We could optimize this by initializing a count before and incrementing it each time our algorithm builds a "true" node.

COMPLEXITY ANALYSIS:

We are building a tree of at most $3^{n + k}$ nodes. The comparisons are constant. I think this means that the time complexity overall is $O(3^n)$, although with memoization this can be reduced to $O(n * k) \in O(n)$ because there are only (n + 1) * (k + 1) distinct problems.
