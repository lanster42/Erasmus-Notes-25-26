### Question 1

Our starting vertex is 1. That means that it's the first discovered. All the other vertices are undiscovered for now. In the first loop we add all the adjacent vertices of vertex 1 to the stack in decending order. That way number 2 is on top of the stack. We then take the top vertex (2) from the stack. Since 2 doesn't have any adjacent vertices, we don't add anything to the stack. In the next loop, we return to vertex one and take the next vertex from the stack (3) and mark it as explored. We then add 6 and 5 to the stack (6 on top). In the next loop we take 6 from the stack and mark it as explored. Since all of its adjacent vertices are already either discovered or explored, we don't add anything to the stack. In the next loop we take 5 from the stack and mark it as explored. SInce 5 doesn't have any adjacent vertices that are unexplored, we don't add anything to the stack. We return to vertex 1, since vertex 4 is the only one left on the stack. In the next loop we take it from the stack and mark it as explored. We have now explored all the vertices and the stack is empty, so the algorithm ends.

```python
from collections import deque

example_graph2 = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['E'],
    'D': ['E'],
    'E': []}

def min_semesters(graph):
    #For this problem we dont need initialization, we only need to remember what is the longest path in the graph
    longest_path = {}
    
    def dfs(u):
        if u in longest_path: #chacking if u is already in longest_path => O(1) because it's a dictionary
            return longest_path[u]
        if not graph[u]: #we add a vertex if it's not already in graph.keys => O(1) because dictionary
            longest_path[u] = 1
            return 1
        
        #This is our dfs
        max_depth = 0
        for adjacent in graph[u]:
            max_depth = max(max_depth, dfs(adjacent)) # with recursion we can always take the longest path that we can from u to all other vertices. This is a DFS recursion, which we already proved that it's O(n)
        longest_path[u] = max_depth + 1
        return longest_path[u]
    
    return max(dfs(u) for u in graph)

print(min_semesters(example_graph2))
```

max() runs DFS once per vertex, but because of longest_path, after the first DFS call on a connected component, all nodes in that component are already computed (saved inside longest_path which has O(1) complexity). So the complexity of this algorithm is still O(n) which is linear.
