```python
import heapq
import math

example_graph2 = {
    'Madrid': {'Rome': 1, 'Amsterdam': 2, 'Berlin': 3},
    'Amsterdam': {'Madrid': 2, 'Berlin': 1},
    'Berlin': {'Amsterdam': 1, 'Madrid': 3, 'Rome': 2, 'Stockholm': 2, 'Kyiv': 3, 'Moscow': 6},
    'Rome': {'Madrid': 1, 'Berlin': 2},
    'Stockholm': {'Berlin': 2, 'Moscow': 3},
    'Kyiv': {'Berlin': 3, 'Moscow': 2},
    'Moscow': {'Berlin': 6, 'Stockholm': 3, 'Kyiv': 2}}


def easiest_path(graph, source_vertex):
    """Note to self: This is NOT just a normal Dijkstra's algorithm since the path will only change when there is a shorter distance and won't change when distances are the same. So it doesn't guarantee that it will take the shortest path first but rather the one it discovers first, which is often the one with more flights. So more thought is needed."""
    
    #Initialization
    length = {u: math.inf for u in graph} #this gives each city infinite length 
    predecessor = {u: None for u in graph}
    flights = {u: math.inf for u in graph} #we will check the number of flights from source with this dict
    length[source_vertex] = 0
    flights[source_vertex] = 0
    
    priority_queue = [(0, 0, source_vertex)] #Dijskstras algorithm uses a priority queue :) (distance, flights, vertex)
    
    #Main loop
    while priority_queue: #creating a loop for each city if queue isnt empty
        current_distance, current_flights, current = heapq.heappop(priority_queue) #taking the first element of the queue, like Q.popleft()
        # Skip if we already found a better path (shorter or equal but with fewer flights)
        if (current_distance > length[current]) or (
            current_distance == length[current] and current_flights > flights[current]
        ):
            continue
        
        for city, weight in graph[current].items(): #This way we can already name both things we need (the city and their distance to current)
            new_distance = current_distance + weight
            new_flights = current_flights + 1
            if (new_distance < length[city]) or (new_distance == length[city] and new_flights < flights[city]): #when the distance is shorter OR the same and less flights
                length[city] = new_distance
                predecessor[city] = current
                flights[city] = new_flights
                heapq.heappush(priority_queue, (new_distance, new_flights, city))
                
    for city in length:
        print(f"- {flights[city]} for {city}.")
        
    return ":)"
            
print(easiest_path(example_graph2, 'Madrid'))
```

```python
example_graph2 = {
    'A': {'B': 1, 'C': 2},
    'B': {'D': 1, 'E': 3},
    'C': {'E': 1},
    'D': {'E': 1},
    'E': {}
}

def dial_shortest_paths(graph, source, W):
    # Initialization
    dist = {v: float('inf') for v in graph}
    dist[source] = 0

    buckets = [[] for _ in range(W + 1)]
    buckets[0].append(source)

    current_index = 0

    while True:
        # Find next non-empty bucket
        while not buckets[current_index]:
            current_index = (current_index + 1) % (W + 1)
            if all(not b for b in buckets):
                return dist

        u = buckets[current_index].pop()

        for v, weight in graph[u].items():
            new_dist = dist[u] + weight
            if new_dist < dist[v]:
                dist[v] = new_dist
                bucket_index = new_dist % (W + 1)
                buckets[bucket_index].append(v)

print(dial_shortest_paths(example_graph2, 'A', 5))
```
