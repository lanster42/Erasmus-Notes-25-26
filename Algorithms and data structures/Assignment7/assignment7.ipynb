### Question 3

First we focus on building the graph. We already have G, S and t. First we add a source vertex s with edges to $s_i$ for all $i \in ({1, 2, \ldots n})$ which have weight $ks_i$ to the corresponding vertex $s_i$ (look at the picture below). This happens in $O(n)$ time. We assign the weights of edges from $s_i$ to $v_i$ and from $v_i$ to $t$ to have weight 1. This happens in $O(|V|)$ time.

All we have to do then is run Ford-Fulkerson's algorithm on the constructed graph and return true if the minimum cut is the sum of $ks_i$. Ford-Fulkerson has time complexity of $O((|E| + |V|) * f)$ which in this example translates to $O((|E| + |S| + 2|V|)ks_i)$ time, which is also the combined complexity.

The main difference is the r paramether which is used to construct the graph before running the Ford-Fulkerson algorithm.

For construction we first need to initiate edges from source $s$ to all clients with capacity 1 ($O(n)$) and edges from base stations to the sink $t$ with capacities $L$ ($O(k)$). This is done so that there are no more than $L$ clients on each base station. For this idea to work, we also need all edges from clients to base stations to have capacities 1, but only to those stations, with which they have less than $r$ distance in between. Calculation of distance is constant. Lastly, we need to initialize the edges from clients so base stations. In worst case scenario each client would have an edge to each base station, making complexity $O(n*k)$.

This would mean that the Ford-Fulkerson would run in $O((nk + n + k)kL)$ time.
