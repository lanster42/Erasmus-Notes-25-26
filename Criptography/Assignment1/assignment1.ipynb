```python
import string
import math

ex1 = "MATHEMATICSFORCRYPTOGRAPHY"
key1 = "LEMON"

def encrypt(plain_txt, key):
    #first we need to construct the key. This isn't the most efficient implementation but will do for now
    rep_count = len(plain_txt) // len(key)
    remainder = len(plain_txt) % len(key)
    new_key = rep_count * key + key[:remainder]
        
    #now the encrypting part
    encrypted_text = ""
    
    for i in range(len(plain_txt)):
        new_index = (ord(plain_txt[i]) + ord(new_key[i]) - 2*ord('A')) % 26
        encrypted_text += chr(new_index + ord('A'))
        
    return encrypted_text

print(encrypt(ex1, key1))
        
```
